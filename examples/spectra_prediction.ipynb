```python
# import neural network
import deepfit_package
from deepfit_package import Net, Structure
model = Net(device='cpu')
```

```python
# import structure
with open('Ni_Salen_.xyz') as f:
  xyz = ''.join(f.readlines()[2:])
  print(xyz)
structure = Structure(xyz, device='cpu')
```

```python
# Preict spectrum from structure
pred_spectrum = model(structure.data)
```

```python
# Visualization
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
sns.set_style("whitegrid")
fig, ax = plt.subplots(figsize=(8, 3), dpi=120)
mask = model.k > 1
ax.plot(model.k[mask], pred_spectrum[0][mask].detach().cpu().numpy(), linewidth=2.5, color='red')
ax.set_xlabel('Wavenumber (k)', fontsize=12), ax.set_ylabel('$\chi(k)k^2$', fontsize=12)
```
